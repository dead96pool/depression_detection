```python
def find_missing_zips(parent_directory):
    expected_range = range(300, 493)
    found_zips = set()

    # Traverse the directory and subdirectory
    for root, _, files in os.walk(parent_directory):
        for file in files:
            if file.endswith('.zip'):
                try:
                    number = int(file.split('_')[0])
                    if 300 <= number <= 492:
                        found_zips.add(number)
                except ValueError:
                    continue

    # Identify missing zips
    missing_zips = [i for i in expected_range if i not in found_zips]

    return missing_zips
```

```python
def analyze_zip_files(parent_directory):
    expected_range = range(300, 493)
    found_zips = set()
    total_size = 0

    # Traverse the directory and subdirectory
    for root, _, files in os.walk(parent_directory):
        for file in files:
            if file.endswith('.zip'):
                try:
                    number = int(file.split('_')[0])
                    if 300 <= number <= 492:
                        found_zips.add(number)
                        # Calculate the size in MB
                        total_size += os.path.getsize(os.path.join(root, file)) / (1024 * 1024)
                except ValueError:
                    continue

    downloaded_zips = sorted(found_zips)
    missing_zips = sorted([i for i in expected_range if i not in found_zips])

    return total_size, downloaded_zips, missing_zips
```

```python
def verify_downloaded_zips(text_file, parent_directory):
    # Read the zip file names and expected sizes from the text file
    expected_sizes = {}
    with open(text_file, 'r') as f:
        next(f)  # Skip the header line
        for line in f:
            line = line.strip()
            if line:
                parts = line.split('\t')
                if len(parts) == 2:
                    name = parts[0]
                    size_str = parts[1]
                    try:
                        size_mb = float(size_str[:-1])  # Remove 'M' and convert to float
                        expected_sizes[name] = size_mb
                    except ValueError:
                        continue

    # Traverse the directory and subdirectory to find downloaded zip files
    found_zips = {}
    for root, _, files in os.walk(parent_directory):
        for file in files:
            if file.endswith('.tar.gz'):
                found_zips[file] = os.path.getsize(os.path.join(root, file)) / (1024 * 1024)  # Convert size to MB

    # Compare downloaded zip sizes with expected sizes
    correct_size_zips = []
    incorrect_size_zips = []

    for name, expected_size in expected_sizes.items():
        if name in found_zips:
            downloaded_size = found_zips[name]
            # Allow a small tolerance for rounding differences (0.5 MB tolerance)
            if abs(downloaded_size - expected_size) <= 0.497:
                correct_size_zips.append(name)
            else:
                incorrect_size_zips.append(name)

    # Identify missing zip files in the text file
    missing_zips = [name for name in expected_sizes.keys() if name not in found_zips]

    return correct_size_zips, incorrect_size_zips, missing_zips
```

```python
# Example usage:
text_file_path = r"file_list_extended.txt"
parent_directory = r'D:\Datasets\zips\E_DAIC'  # Replace with your parent directory path

correct_size_zips, incorrect_size_zips, missing_zips = verify_downloaded_zips(text_file_path, parent_directory)

print("Correct size zip files:", correct_size_zips)
print("Incorrect size zip files:", incorrect_size_zips)
print("Missing zip files from text file:", missing_zips)
```

```python
def calculate_total_sizes(parent_directory):
    total_downloaded_size = 0

    # Traverse the directory and subdirectory to find downloaded zip files
    for root, _, files in os.walk(parent_directory):
        for file in files:
            if file.endswith('.tar.gz'):
                file_path = os.path.join(root, file)
                total_downloaded_size += os.path.getsize(file_path) / (1024 * 1024)  # Convert size to MB

    return total_downloaded_size

def calculate_remaining_size(text_file, parent_directory):
    expected_sizes = {}
    total_expected_size = 0

    # Read the zip file names and expected sizes from the text file
    with open(text_file, 'r') as f:
        next(f)  # Skip the header line
        for line in f:
            line = line.strip()
            if line:
                parts = line.split('\t')
                if len(parts) == 2:
                    name = parts[0]
                    size_str = parts[1]
                    try:
                        size_mb = float(size_str[:-1])  # Remove 'M' and convert to float
                        expected_sizes[name] = size_mb
                        total_expected_size += size_mb
                    except ValueError:
                        continue

    # Calculate total downloaded size
    total_downloaded_size = calculate_total_sizes(parent_directory)

    # Calculate remaining size to download
    remaining_size_to_download = total_expected_size - total_downloaded_size

    return total_downloaded_size, remaining_size_to_download
```

```python
# Example usage:
text_file_path = r"file_list_extended.txt"  # Replace with your text file path
parent_directory = r'D:\Datasets\zips\E_DAIC'  # Replace with your parent directory path

total_downloaded_size, remaining_size_to_download = calculate_remaining_size(text_file_path, parent_directory)

print(f"Total size of downloaded zip files: {total_downloaded_size:.2f} MB")
print(f"Total size of zip files remaining to download: {remaining_size_to_download:.2f} MB")
```
