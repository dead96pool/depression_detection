# official training split

```python
training_labels = pd.read_csv(r"datasets\DAIC-Woz\train_split_Depression_AVEC2017.csv")
training_ids = list(training_labels["Participant_ID"])
display(training_labels)
print(training_ids)
```

```python
# training csv audio
training_audio = pd.DataFrame()

for id in training_ids:
    print(id)
    temp = pd.read_csv(fr"audio_matchingrows\{id}.csv")

    training_audio = pd.concat([training_audio, temp], ignore_index = True, axis = 0)
    print(training_audio.shape)
```

```python
display(training_audio)
```

```python
training_audio["Target"].value_counts()
```

```python
training_audio.to_csv("training_audio.csv", index = False)
```
