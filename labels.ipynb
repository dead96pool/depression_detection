```python
import pandas as pd
```

```python
dev_df = pd.read_csv(r"datasets\DAIC-Woz\dev_split_Depression_AVEC2017.csv")
train_df = pd.read_csv(r"datasets\DAIC-Woz\train_split_Depression_AVEC2017.csv")
test_df = pd.read_csv(r"datasets\DAIC-Woz\full_test_split.csv")
```

```python
#print(test_df.columns)
```

```python
dev_new = dev_df[["Participant_ID", "PHQ8_Binary", "PHQ8_Score"]]
train_new = train_df[["Participant_ID", "PHQ8_Binary", "PHQ8_Score"]]
test_new = test_df[["Participant_ID", "PHQ_Binary", "PHQ_Score"]]
test_new.columns = ["Participant_ID", "PHQ8_Binary", "PHQ8_Score"]
```

```python
PHQ8_labels = pd.concat([dev_new, train_new, test_new], axis = 0)
PHQ8_labels.sort_values("Participant_ID", inplace = True)
PHQ8_labels
```

```python
# saving the labels
PHQ8_labels.to_csv("PHQ8_labels.csv", index = False)
```
