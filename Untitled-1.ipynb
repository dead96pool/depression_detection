```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
test = pd.read_csv(r"datasets\DAIC-Woz\full_test_split.csv")
train = pd.read_csv(r"datasets\DAIC-Woz\train_split_Depression_AVEC2017.csv")
dev = pd.read_csv(r"datasets\DAIC-Woz\dev_split_Depression_AVEC2017.csv")

print(test.shape)
print(train.shape)
print(dev.shape)
```

```python
train.drop(columns = ['PHQ8_Score', 'Gender',
       'PHQ8_NoInterest', 'PHQ8_Depressed', 'PHQ8_Sleep', 'PHQ8_Tired',
       'PHQ8_Appetite', 'PHQ8_Failure', 'PHQ8_Concentrating', 'PHQ8_Moving'], inplace = True)
train.columns
```

```python
dev.drop(columns = ['PHQ8_Score', 'Gender',
       'PHQ8_NoInterest', 'PHQ8_Depressed', 'PHQ8_Sleep', 'PHQ8_Tired',
       'PHQ8_Appetite', 'PHQ8_Failure', 'PHQ8_Concentrating', 'PHQ8_Moving'], inplace = True)
dev.columns
```

```python
test.drop(columns = ['PHQ_Score', 'Gender'], inplace = True)
test.columns = ['Participant_ID', 'PHQ8_Binary']
```

```python
combined = pd.concat([test, train, dev], ignore_index=True, axis = 0)
combined.shape
```

```python
labels = []

labels = list(train["PHQ8_Binary"]) + list(dev["PHQ8_Binary"]) + list(test["PHQ_Binary"])


print(len(labels))
print(labels)
```

```python
class_counts = combined["PHQ8_Binary"].value_counts()
class_counts
```

```python
plt.figure(figsize=(8, 6))
plt.pie(class_counts, labels=["Not-Depressed", "Depressed"], autopct='%1.1f%%', colors=sns.color_palette('viridis', len(class_counts)))
plt.title('Class Distribution')
plt.show()
```

```python
colors = sns.color_palette('viridis', len(class_counts))
explode = [0.1] * len(class_counts)

plt.figure(figsize=(6, 4))
plt.pie(class_counts, 
        #labels=["Not-Depressed", "Depressed"], 
        autopct='%1.1f%%', 
        colors=colors,
        wedgeprops=dict(width=0.7, edgecolor='w'))

plt.title('Class Distribution', fontsize=16, weight='bold')

# Add a legend
plt.legend(["Not-Depressed", "Depressed"], title='Classes', loc='upper left', bbox_to_anchor=(1, 1))

# Display the plot
plt.tight_layout()  # Adjust layout to make room for the legend
plt.show()
```
