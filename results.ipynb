```python
import numpy as np
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
def plot_confusion_matrix(model_name, cm):
    #cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f'Confusion Matrix for {model_name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()
```

```python
models = {
            "rf_video":   ["Random Forest (video modality)", np.array([[24478, 440], [7701, 3498]])],
            "ffnn_video": ["FFNN (video modality)", np.array([[21366, 3536], [4412, 6803]])],
            "bilstm_audio": ["BiLSTM + Attention (audio modality)", np.array([[90000, 5476], [11244, 28958]])],
            "lgbm_video": ["LightGBM (video modality)", np.array([[23575, 1327], [8810, 2405]])],
            "bilstm_mm":  ["BiLSTM + Attention (Multimodal)", np.array([[2430, 182], [65, 1112]])]

}
```

```python
# plotting cm
plot_confusion_matrix(models["rf_video"][0], models["rf_video"][1])
```

```python
print(models["rf_video"][0])
print(models["rf_video"][1])
```

```python
import math
def print_results(model):
    model_name = models[model][0]
    cm = models[model][1]
    tn, fp, fn, tp = cm.ravel()

    #print(model_name)
    #print(cm)
    acc = (tp + tn) / (tp + tn + fp + fn)
    pre = tp / (tp + fp)
    rec = tp / (tp + fn)
    f1 = (2*tp) / ((2*tp) + fp + fn)
    spe = tn / (tn + fp)
    npv = tn / (tn + fn)
    fnr = fn / (fn + tp)
    #mcc = ((tp*tn) - (fp*fn)) / math.sqrt((tp+fp)*(tp+fn)*(tn+fp)*(tn+fn))

    print(f"{model_name}:\n")
    print(f"Accuracy: {acc:0.4f}")
    print(f"Precision: {pre:0.4f}")
    print(f"Recall: {rec:0.4f}")
    print(f"F1 Score: {f1:0.4f}")
    print(f"Specificity: {spe:0.4f}")
    print(f"Negative Predictive Value: {npv:0.4f}")
    print(f"ROC-AUC:")
    print(f"False Negative Rate: {fnr:0.4f}")
    #print(f"Mathews Correlation Coefficient: {mcc:0.4f}")
```

```python
print_results("rf_video")
```
